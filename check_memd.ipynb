```python
from memd import memd
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
```

```python

path = 'data/GBPUSD/minutes.csv'
df = pd.read_csv(path)[4000000:4000060]
```

```python
cols = ['open', 'high', 'low', 'close']
signals = []
for col in ['open', 'high', 'low', 'close']:
    signals.append(np.array(df[col]))
signal = np.array(df[cols])
```

```python
imfs = memd(signal, seq=4)
```

```python
num_imfs = imfs.shape[0]

```

```python
fig, axes = plt.subplots(num_imfs+1, 1, figsize=(10, 2 * num_imfs), sharex=True)

axes[0].plot(np.array(df['open']), label='Open', color='blue')
axes[0].plot(np.array(df['high']), label='High', color='green')
axes[0].plot(np.array(df['low']), label='Low', color='red')
axes[0].plot(np.array(df['close']), label='Close', color='orange')
axes[0].set_title(f'Signal')
axes[0].legend()
# Plot each IMF
for i in range(num_imfs):
    axes[i+1].plot(imfs[i][0], label='Open', color='blue')
    axes[i+1].plot(imfs[i][1], label='High', color='green')
    axes[i+1].plot(imfs[i][2], label='Low', color='red')
    axes[i+1].plot(imfs[i][3], label='Close', color='orange')
    axes[i+1].set_title(f'IMF {i+1}')
    axes[i+1].legend()

plt.xlabel('Time (minutes)')
plt.tight_layout()
plt.show()
```
