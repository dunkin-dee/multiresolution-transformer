```python
import os
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('recent_data/BTCUSD/minutes.csv')
```

```python
split_point = int(0.9*len(df))
```

```python
df = df[split_point:]
```

```python
len(df)
```

```python
train_data, temp_data = train_test_split(df, test_size=0.30, shuffle=False)
```

```python
val_data, test_data = train_test_split(temp_data, test_size=0.50, shuffle=False)
```

```python
folder_path = "recent_data/BTCUSD"
```

```python
train_data.to_csv(os.path.join(folder_path, 'train.csv'), index=False)
val_data.to_csv(os.path.join(folder_path, 'val.csv'), index=False)
test_data.to_csv(os.path.join(folder_path, 'test.csv'), index=False)
```
